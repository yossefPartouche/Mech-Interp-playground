## Positional Embedding

- We have second learnable paramter that enables to model to learn about the importance of the order of the tokens. 


### Useful techniques 
  
``torch.arange()`` - Produces a ``tensor([0, 1, 2, 3...]), shape (T,)``

We embed the postional information the embedded tokens so if we have a shape of ``(_, T, C)`` then for each ``T`` we have ``C`` positonal values to insert

### Toy Example

```python
import torch
import torch.nn as nn

wpe = nn.Embedding(4, 4)   # (T, C) = (4, 4)

print(wpe.weight.data)

# Overwrite random init with our toy W_P from 2.2, so we can hand-check.
wpe.weight.data = torch.tensor([
    [0.01, 0.02, 0.03, 0.04],   # row 0 (pos 0)
    [0.05, 0.06, 0.07, 0.08],   # row 1 (pos 1)
    [0.09, 0.10, 0.11, 0.12],   # row 2 (pos 2)
    [0.13, 0.14, 0.15, 0.16],   # row 3 (pos 3)
])
# change the value between 0, 1, 2, 3, 4, 5 see what happens

idx = torch.arange(3)
print(idx)

some = wpe(idx)
print(some)

#pos_emb = wpe(idx)

#print(pos_emb)

#print(pos_emb.shape)
```

### Implementing Positional Embedding

```python
class PositionEmbedding(nn.Module):
    def __init__(self, block_size, n_embd):
        super().__init__()
        self.wpe = nn.Embedding(block_size, n_embd)

    def forward(self, idx):
        B, T = idx.shape
        pos = torch.arange(T)
        pos_emb = self.wpe(pos)
        return pos_emb
```

```python
pe = PositionEmbedding(4, 4)
out = pe(torch.tensor([[0, 1, 2, 3]]))   # (B, T) in
print(out.shape)                          # torch.Size([4, 4]) = (T, C)
```
