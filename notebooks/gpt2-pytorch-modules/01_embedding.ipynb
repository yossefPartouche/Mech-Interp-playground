## Components of Embedding 

- ``init()`` is where you place the learnable parameters
- ``nn.Embedding`` - module holding learnable parameters of matrix ``(V,C)`` 
- ``nn.Embedding.weight`` - returns the weight of the parameters


```python
import torch 
import torch.nn as nn
```

### Toy Example

```python
wte = nn.Embedding(6, 4)

# Normally these are random but this also shows how to manually add values 
wte.weight.data = torch.tensor([
    [0.1, 0.2, 0.3, 0.4],   # row 0 (the)
    [0.5, 0.6, 0.7, 0.8],   # row 1 (cat)
    [0.9, 1.0, 1.1, 1.2],   # row 2 (sat)
    [1.3, 1.4, 1.5, 1.6],   # row 3 (on)
    [1.7, 1.8, 1.9, 2.0],   # row 4 (mat)
    [2.1, 2.2, 2.3, 2.4],   # row 5 (<eos>)
])

# This is part of the Embedding module
print(wte)
# This is part of the torch module which has the ".shape" functionality 
print(wte.weight.shape)
# this is each row 
print(wte(torch.tensor([0,1])))

idx = torch.tensor([[0, 1, 2, 3]]) 

print("This vector has shape:", idx.shape)

print(wte(idx))
```

### Implementing Token Embedding class

```python
class TokenEmbedding(nn.Module):
    def __init__(self, vocab_size, n_embd):
        super().__init__()
        self.wte = nn.Embedding(vocab_size, n_embd)
    def forward(self, idx):
        return self.wte(idx)
```

### Calling 

```python
te = TokenEmbedding(6, 4)

print(te(torch.tensor([[0,1,2,3]])).shape)
```
